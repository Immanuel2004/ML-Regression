```python
import sys
import pandas as pd
from src.logger import logging
from src.exception import CustomException

def upload_file() -> pd.DataFrame:
    try:
        file_path = input("Enter the filename (with path if not in current directory): ").strip()
        logging.info(f"User provided dataset path: {file_path}")

        df = pd.read_csv(file_path)  # Read the CSV file
        logging.info("Dataset successfully loaded into a DataFrame.")
        return df

    except FileNotFoundError:
        logging.error("The specified file was not found.")
        raise CustomException("File not found. Please check the path and try again.", sys)
    except pd.errors.EmptyDataError:
        logging.error("The provided file is empty or invalid.")
        raise CustomException("The provided file is empty or invalid.", sys)
    except Exception as e:
        logging.error(f"Unexpected error: {e}")
        raise CustomException(e, sys)
```
